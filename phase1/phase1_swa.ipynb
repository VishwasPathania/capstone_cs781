we define the cleaning rules from the data audit and test them on a small sample before cleaning the full corpus

```python
import re
import unicodedata

def clean_text(text):
    text=unicodedata.normalize("NFKC",str(text))
    text=text.replace("\r\n","\n").replace("\r","\n")
    text=text.replace("\x00"," ")
    text=text.replace("\ufffd"," ")

    text=re.sub(r"(?im)^\s*Signature Not Verified\s*$","",text)
    text=re.sub(r"(?im)^\s*Digitally signed by.*$","",text)

    text=re.sub(r"[^\S\n]+"," ",text)
    text=re.sub(r"\n[ \t]+","\n",text)
    text=re.sub(r"\n{3,}","\n\n",text)

    return text.strip()

for name,path in paths.items():

    print("\n",name)

    shown=0

    with open(path,"rb") as f:
        for record in ijson.items(f,"item"):

            raw_text=str(record.get("text",""))
            cleaned_text=clean_text(raw_text)

            if raw_text!=cleaned_text:

                print("original:")
                print(raw_text[:500])

                print("\ncleaned:")
                print(cleaned_text[:500])

                print("-"*80)

                shown+=1

            if shown>=3:
                break
```

we clean the full legal corpus using the rules tested above and remove exact duplicate documents while keeping the original metadata

```python
import ijson
import json
import os
import hashlib
import re
import unicodedata

clean_dir="clean_corpus"
os.makedirs(clean_dir,exist_ok=True)

clean_stats={}

for name,path in paths.items():

    output_name=name.replace(".json",".jsonl")
    output_path=os.path.join(clean_dir,output_name)

    seen=set()
    total=0
    kept=0
    duplicates=0
    empty=0

    with open(path,"rb") as f,open(output_path,"w",encoding="utf-8") as out:

        for record in ijson.items(f,"item"):

            total+=1

            raw_text=record.get("text","")
            text=unicodedata.normalize("NFKC",str(raw_text))

            text=text.replace("\r\n","\n").replace("\r","\n")
            text=text.replace("\x00"," ")
            text=text.replace("\ufffd"," ")

            text=re.sub(
                r"(?im)^\s*Signature Not Verified\s*$",
                "",
                text
            )

            text=re.sub(
                r"(?im)^\s*Digitally signed by.*$",
                "",
                text
            )

            text=re.sub(r"[^\S\n]+"," ",text)
            text=re.sub(r"\n[ \t]+","\n",text)
            text=re.sub(r"\n{3,}","\n\n",text)

            text=text.strip()

            if not text:
                empty+=1
                continue

            text_hash=hashlib.sha256(
                text.encode("utf-8")
            ).digest()

            if text_hash in seen:
                duplicates+=1
                continue

            seen.add(text_hash)

            record["text"]=text

            out.write(
                json.dumps(
                    record,
                   ensure_ascii=False
                )+"\n"
            )

            kept+=1

    clean_stats[name]={
        "total":total,
        "kept":kept,
        "duplicates":duplicates,
        "empty_after_cleaning":empty
    }

for name,stats_ in clean_stats.items():

    print("\n",name)
    print("original documents:",stats_["total"])
    print("kept documents:",stats_["kept"])
    print("duplicates removed:",stats_["duplicates"])
    print("empty documents removed:",stats_["empty_after_cleaning"])
```

we split the cleaned documents at the document level into fixed train and validation sets so validation text never leaks into tokenizer training


```python
import os
import json
import random

split_dir="data_split"
os.makedirs(split_dir,exist_ok=True)

seed=42
val_fraction=0.02

random.seed(seed)

split_stats={}

for name in clean_stats:

    input_name=name.replace(".json",".jsonl")
    input_path=os.path.join(clean_dir,input_name)

    train_name=input_name.replace(".jsonl","_train.jsonl")
    val_name=input_name.replace(".jsonl","_val.jsonl")

    train_path=os.path.join(split_dir,train_name)
    val_path=os.path.join(split_dir,val_name)

    records=[]

    with open(input_path,"r",encoding="utf-8") as f:
        for line in f:
            records.append(json.loads(line))

    random.shuffle(records)

    val_count=int(len(records)*val_fraction)

    val_records=records[:val_count]
    train_records=records[val_count:]

    with open(train_path,"w",encoding="utf-8") as f:
        for record in train_records:
            f.write(json.dumps(record,ensure_ascii=False)+"\n")

    with open(val_path,"w",encoding="utf-8") as f:
        for record in val_records:
            f.write(json.dumps(record,ensure_ascii=False)+"\n")

    split_stats[name]={
        "total":len(records),
        "train":len(train_records),
        "validation":len(val_records)
    }

for name,stats_ in split_stats.items():
    print("\n",name)
    print("total:",stats_["total"])
    print("train:",stats_["train"])
    print("validation:",stats_["validation"])
```

we now define the exact scratch byte-level bpe tokenizer from assignment 1 that will be used for all tokenizer experiments

```python
from collections import Counter,defaultdict
import heapq
import re

SPECIAL_TOKENS=["<pad>","<unk>","<s>","</s>"]
N_SPECIAL=4
N_BYTES=256
BASE_VOCAB_SIZE=N_SPECIAL+N_BYTES

WORD_RE=re.compile(r"\S+")


def _display_byte(b):
    if 32<=b<=126 and b!=92:
        return chr(b)
    return f"<0x{b:02X}>"


def _display_token(data):
    try:
        text=data.decode("utf-8")
        if all(ch.isprintable() or ch in "\n\t\r" for ch in text):
            return text
    except UnicodeDecodeError:
        pass

    if len(data)==1:
        return _display_byte(data[0])

    return "<bytes:"+data.hex()+">"


class A1BPETokenizer:

    def __init__(self,itos,id_to_bytes,merges):
        self.itos=list(itos)
        self._id_to_bytes=list(id_to_bytes)
        self.merges=[tuple(m) for m in merges]
        self.vocab_size=len(self.itos)

        self.stoi={s:i for i,s in enumerate(self.itos)}

        self._byte_to_id={
            bytes([b]):N_SPECIAL+b
            for b in range(256)
        }

        self._merge_ranks={
            pair:r
            for r,pair in enumerate(self.merges)
        }


    @staticmethod
    def _pair_lex_key(pair,id_to_bytes):
        a,b=pair
        return(id_to_bytes[a],id_to_bytes[b],a,b)


    @classmethod
    def train(cls,train_text,vocab_size):

        if not isinstance(train_text,str):
            raise TypeError("train_text must be a string")

        if vocab_size<BASE_VOCAB_SIZE:
            raise ValueError(
                f"vocab_size must be >= {BASE_VOCAB_SIZE}; "
                "this includes 4 reserved tokens + 256 byte symbols."
            )

        word_freq=Counter()

        for m in WORD_RE.finditer(train_text):
            word=m.group(0).encode("utf-8")

            if word:
                word_freq[word]+=1


        id_to_bytes=[b""]*N_SPECIAL

        id_to_bytes.extend(
            bytes([b])
            for b in range(256)
        )


        itos=list(SPECIAL_TOKENS)

        itos.extend(
            _display_byte(b)
            for b in range(256)
        )


        words=[]
        freqs=[]

        byte_to_id={
            bytes([b]):N_SPECIAL+b
            for b in range(256)
        }


        for word,freq in word_freq.items():

            words.append(
                [
                    byte_to_id[bytes([b])]
                    for b in word
                ]
            )

            freqs.append(freq)


        merges=[]

        if vocab_size==BASE_VOCAB_SIZE or not words:
            return cls(
                itos,
                id_to_bytes,
                merges
            )


        pair_counts={}
        pair_words=defaultdict(set)


        def sequence_pairs(seq):
            return Counter(
                (seq[i],seq[i+1])
                for i in range(len(seq)-1)
            )


        for wid,seq in enumerate(words):

            local=sequence_pairs(seq)
            f=freqs[wid]

            for pair,occ in local.items():

                pair_counts[pair]=(
                    pair_counts.get(pair,0)+f*occ
                )

                pair_words[pair].add(wid)


        heap=[]

        for pair,count in pair_counts.items():

            heapq.heappush(
                heap,
                (
                    -count,
                    cls._pair_lex_key(
                        pair,
                        id_to_bytes
                    ),
                    pair
                )
            )


        while len(itos)<vocab_size and heap:

            best_pair=None
            best_count=0


            while heap:

                neg_count,_,pair=heapq.heappop(heap)

                current=pair_counts.get(pair,0)

                if current!=-neg_count:
                    continue

                best_pair=pair
                best_count=current
                break


            if best_pair is None or best_count<2:
                break


            a,b=best_pair

            new_id=len(itos)

            id_to_bytes.append(
                id_to_bytes[a]+id_to_bytes[b]
            )

            itos.append(
                _display_token(
                    id_to_bytes[new_id]
                )
            )

            merges.append((a,b))


            affected=list(
                pair_words.get(
                    best_pair,
                    set()
                )
            )


            for wid in affected:

                old_seq=words[wid]

                old_pairs=sequence_pairs(old_seq)

                f=freqs[wid]


                for pair,occ in old_pairs.items():

                    pair_counts[pair]-=f*occ

                    pair_words[pair].discard(wid)

                    if pair_counts[pair]<=0:

                        pair_counts.pop(pair,None)
                        pair_words.pop(pair,None)


                new_seq=[]
                i=0

                while i<len(old_seq):

                    if(
                        i+1<len(old_seq)
                        and old_seq[i]==a
                        and old_seq[i+1]==b
                    ):

                        new_seq.append(new_id)
                        i+=2

                    else:

                        new_seq.append(old_seq[i])
                        i+=1


                words[wid]=new_seq

                new_pairs=sequence_pairs(new_seq)


                for pair,occ in new_pairs.items():

                    pair_counts[pair]=(
                        pair_counts.get(pair,0)+f*occ
                    )

                    pair_words[pair].add(wid)


                changed=set(old_pairs)|set(new_pairs)


                for pair in changed:

                    count=pair_counts.get(pair,0)

                    if count>0:

                        heapq.heappush(
                            heap,
                            (
                                -count,
                                cls._pair_lex_key(
                                    pair,
                                    id_to_bytes
                                ),
                                pair
                            )
                        )


            pair_counts.pop(best_pair,None)
            pair_words.pop(best_pair,None)


        tokenizer=cls(
            itos,
            id_to_bytes,
            merges
        )

        assert tokenizer.vocab_size==len(tokenizer.itos)
        assert tokenizer.vocab_size==len(tokenizer._id_to_bytes)

        return tokenizer


    def get_vocab_size(self):
        return self.vocab_size


    def _encode_word(self,word):

        raw=word.encode("utf-8")

        seq=[
            self._byte_to_id[bytes([b])]
            for b in raw
        ]

        if len(seq)<2 or not self._merge_ranks:
            return seq


        n=len(seq)

        prev=[-1]+list(range(0,n-1))
        nxt=list(range(1,n))+[-1]
        active=[True]*n


        heap=[]

        for i in range(n-1):

            rank=self._merge_ranks.get(
                (seq[i],seq[i+1])
            )

            if rank is not None:

                heapq.heappush(
                    heap,
                    (
                        rank,
                        i,
                        i+1
                    )
                )


        while heap:

            rank,left,right=heapq.heappop(heap)


            if not active[left] or not active[right]:
                continue

            if nxt[left]!=right or prev[right]!=left:
                continue

            if self._merge_ranks.get(
                (seq[left],seq[right])
            )!=rank:
                continue


            merged_id=BASE_VOCAB_SIZE+rank

            seq[left]=merged_id
            active[right]=False


            right_next=nxt[right]

            nxt[left]=right_next

            if right_next!=-1:
                prev[right_next]=left


            left_prev=prev[left]

            if left_prev!=-1:

                pair=(
                    seq[left_prev],
                    seq[left]
                )

                new_rank=self._merge_ranks.get(pair)

                if new_rank is not None:

                    heapq.heappush(
                        heap,
                        (
                            new_rank,
                            left_prev,
                            left
                        )
                    )


            if right_next!=-1:

                pair=(
                    seq[left],
                    seq[right_next]
                )

                new_rank=self._merge_ranks.get(pair)

                if new_rank is not None:

                    heapq.heappush(
                        heap,
                        (
                            new_rank,
                            left,
                            right_next
                        )
                    )


        out=[]

        i=0

        while i!=-1:

            if active[i]:
                out.append(seq[i])

            i=nxt[i]

        return out


    def encode(self,text):

        if not isinstance(text,str):
            raise TypeError("text must be a string")

        ids=[]
        cursor=0


        for m in WORD_RE.finditer(text):

            if m.start()>cursor:

                ws=text[cursor:m.start()].encode("utf-8")

                ids.extend(
                    N_SPECIAL+b
                    for b in ws
                )


            ids.extend(
                self._encode_word(
                    m.group(0)
                )
            )

            cursor=m.end()


        if cursor<len(text):

            ws=text[cursor:].encode("utf-8")

            ids.extend(
                N_SPECIAL+b
                for b in ws
            )


        return ids


    def decode(self,ids):

        chunks=[]
        byte_buffer=bytearray()


        def flush():

            if byte_buffer:

                chunks.append(
                    bytes(byte_buffer).decode(
                        "utf-8",
                        errors="strict"
                    )
                )

                byte_buffer.clear()


        for idx in ids:

            idx=int(idx)

            if not 0<=idx<self.vocab_size:

                raise ValueError(
                    f"token id {idx} outside vocabulary "
                    f"[0,{self.vocab_size})"
                )


            if idx<N_SPECIAL:

                flush()
                chunks.append(
                    SPECIAL_TOKENS[idx]
                )

            else:

                byte_buffer.extend(
                    self._id_to_bytes[idx]
                )


        flush()

        return "".join(chunks)


    def save_json(self,path):

        import json

        with open(
            path,
            "w",
            encoding="utf-8"
        ) as f:

            json.dump(
                {
                    "merges":self.merges,
                    "itos":self.itos,
                    "vocab_size":self.vocab_size
                },
                f,
                ensure_ascii=False,
                indent=2
            )


def train_tokenizer(train_text,vocab_size):
    return A1BPETokenizer.train(
        train_text,
        vocab_size
    )
```

we now create a 500m-word(20%) representative tokenizer sample from the cleaned training files while preserving the natural source proportions

```python
import os
import json
import random
from tqdm.auto import tqdm

split_dir="data_split"

source_files={
    "acts":os.path.join(split_dir,"train_acts_train.jsonl"),
    "high_courts":os.path.join(split_dir,"train_high_courts_train.jsonl"),
    "supreme_court":os.path.join(split_dir,"train_supreme_court_train.jsonl"),
    "tribunals":os.path.join(split_dir,"train_tribunals_train.jsonl")
}

total_target_words=500_000_000

# ---------------------------------------------------------------
# pass 1: count words in each source
# ---------------------------------------------------------------
source_word_counts={}

for source,path in source_files.items():

    total_words=0

    print("\ncounting:",source)

    with open(path,"r",encoding="utf-8") as f:

        for line in tqdm(f,desc=source):

            record=json.loads(line)
            text=record.get("text","")

            total_words+=sum(
                1 for _ in WORD_RE.finditer(text)
            )

    source_word_counts[source]=total_words

print("\nsource word counts:")

for source,count in source_word_counts.items():
    print(source,":",count)

total_words=sum(source_word_counts.values())

print("\ntotal training words:",total_words)

# ---------------------------------------------------------------
# natural-proportion targets
# ---------------------------------------------------------------
source_targets={
    source:int(
        total_target_words*count/total_words
    )
    for source,count in source_word_counts.items()
}

print("\ntarget tokenizer words:")

for source,target in source_targets.items():
    print(source,":",target)

# ---------------------------------------------------------------
# pass 2: randomly sample documents from each source
# ---------------------------------------------------------------
random.seed(42)

os.makedirs("tokenizer_sample_500m",exist_ok=True)

sample_path=os.path.join(
    "tokenizer_sample_500m",
    "tokenizer_train_sample_500m.txt"
)

with open(sample_path,"w",encoding="utf-8") as out:

    for source,path in source_files.items():

        target=source_targets[source]
        selected_words=0

        source_probability=target/source_word_counts[source]

        print(
            "\nsampling:",
            source,
            "| target words:",
            target,
            "| probability:",
            round(source_probability,6)
        )

        with open(path,"r",encoding="utf-8") as f:

            for line in tqdm(f,desc=source):

                if selected_words>=target:
                    break

                if random.random()>source_probability:
                    continue

                record=json.loads(line)
                text=record.get("text","")

                matches=list(WORD_RE.finditer(text))
                doc_words=len(matches)

                if doc_words==0:
                    continue

                remaining=target-selected_words

                if doc_words<=remaining:

                    out.write(text)
                    out.write("\n")

                    selected_words+=doc_words

                else:

                    end_position=matches[remaining-1].end()

                    out.write(
                        text[:end_position]
                    )
                    out.write("\n")

                    selected_words+=remaining

        print(
            source,
            "selected words:",
            selected_words
        )

print("\nsample saved:",sample_path)

sample_size_gb=(
    os.path.getsize(sample_path)/(1024**3)
)

print(
    "sample size (gb):",
    round(sample_size_gb,3)
)
```

```python
import os
import json
import numpy as np

with open(
    os.path.join(shard_root,"manifest.json"),
    "r",
    encoding="utf-8"
) as f:
    manifest=json.load(f)

sources=[
    "acts",
    "high_courts",
    "supreme_court",
    "tribunals"
]

context_length=1024
stride=1024

for split in ["train","val"]:

    print("\n",split)

    for source in sources:

        source_entries=[
            (manifest_id,entry)
            for manifest_id,entry in enumerate(manifest)
            if entry["split"]==split
            and entry["source"]==source
        ]

        index_parts=[]
        total_sequences=0

        for manifest_id,entry in source_entries:

            shard_tokens=entry["tokens"]

            max_start=(
                shard_tokens
                -context_length
                -1
            )

            if max_start<0:
                continue

            starts=np.arange(
                0,
                max_start+1,
                stride,
                dtype=np.uint32
            )

            if len(starts)==0:
                continue

            shard_ids=np.full(
                len(starts),
                manifest_id,
                dtype=np.uint32
            )

            shard_index=np.column_stack(
                (shard_ids,starts)
            )

            index_parts.append(shard_index)
            total_sequences+=len(starts)

        if index_parts:

            index=np.concatenate(
                index_parts,
                axis=0
            )

        else:

            index=np.empty(
                (0,2),
                dtype=np.uint32
            )

        index_path=os.path.join(
            shard_root,
            f"{split}_{source}_sequence_index.npy"
        )

        np.save(
            index_path,
            index
        )

        print(
            source,
            "| sequences:",len(index),
            "| shape:",index.shape,
            "| saved:",index_path
        )

print("\nsource-specific sequence indices created")
```

```python
import numpy as np
from torch.utils.data import Dataset

class MixedLegalDataset(Dataset):

    def __init__(
        self,
        train_datasets,
        mixing_strategy="natural",
        alpha=0.5,
        seed=42
    ):
        self.train_datasets=train_datasets
        self.sources=list(train_datasets.keys())
        self.alpha=alpha
        self.rng=np.random.default_rng(seed)

        self.indices={
            source:np.arange(
                len(train_datasets[source]),
                dtype=np.uint32
            )
            for source in self.sources
        }

        self.source_ids={
            source:i
            for i,source in enumerate(self.sources)
        }

        self.mixing_strategy=mixing_strategy

        self.sequence_plan=self._build_plan()

    def _build_plan(self):

        lengths=np.array(
            [
                len(self.indices[source])
                for source in self.sources
            ],
            dtype=np.float64
        )

        if self.mixing_strategy=="natural":

            counts=lengths.astype(np.int64)

        elif self.mixing_strategy=="equal_downsample":

            n=int(lengths.min())

            counts=np.full(
                len(self.sources),
                n,
                dtype=np.int64
            )

        elif self.mixing_strategy=="smoothed":

            weights=lengths**self.alpha
            probabilities=weights/weights.sum()

            max_total=int(
                np.floor(
                    np.min(
                        lengths/probabilities
                    )
                )
            )

            counts=np.floor(
                probabilities*max_total
            ).astype(np.int64)

        else:

            raise ValueError(
                "mixing_strategy must be natural,smoothed,or equal_downsample"
            )

        plan_parts=[]

        for source,count in zip(
            self.sources,
            counts
        ):

            count=int(count)

            if count>len(self.indices[source]):

                raise ValueError(
                    f"{source} requires {count} sequences "
                    f"but only {len(self.indices[source])} are available"
                )

            selected=self.rng.choice(
                self.indices[source],
                size=count,
                replace=False
            )

            source_id=self.source_ids[source]

            source_part=np.column_stack(
                (
                    np.full(
                        count,
                        source_id,
                        dtype=np.uint32
                    ),
                    selected
                )
            )

            plan_parts.append(source_part)

        plan=np.concatenate(
            plan_parts,
            axis=0
        )

        self.rng.shuffle(plan)

        self.target_counts={
            source:int(count)
            for source,count in zip(
                self.sources,
                counts
            )
        }

        self.source_probabilities={
            source:float(
                count/len(plan)
            )
            for source,count in zip(
                self.sources,
                counts
            )
        }

        return plan

    def __len__(self):
        return len(self.sequence_plan)

    def __getitem__(self,i):

        source_id=int(
            self.sequence_plan[i,0]
        )

        local_index=int(
            self.sequence_plan[i,1]
        )

        source=self.sources[source_id]

        return self.train_datasets[
            source
        ][local_index]
```

```python
#cell 87
#RoPE from Assignment 3

def build_rope_cache(
    max_seq_len,
    d_head,
    base=10000.0
):
    assert d_head % 2 == 0

    inv_freq=(
        1.0/
        (
            base**(
                torch.arange(
                    0,
                    d_head//2,
                    dtype=torch.float32
                )*2/d_head
            )
        )
    )

    positions=torch.arange(
        max_seq_len,
        dtype=torch.float32
    )

    freqs=torch.outer(
        positions,
        inv_freq
    )

    angles=torch.cat(
        [freqs,freqs],
        dim=-1
    )

    return angles.cos(),angles.sin()


def rotate_half(x):
    x1,x2=torch.chunk(
        x,
        2,
        dim=-1
    )

    return torch.cat(
        [-x2,x1],
        dim=-1
    )


def apply_rope(
    x,
    cos,
    sin
):
    #keep the RoPE tensors in the same dtype/device
    #as x during mixed-precision training
    cos=cos.to(
        device=x.device,
        dtype=x.dtype
    )

    sin=sin.to(
        device=x.device,
        dtype=x.dtype
    )

    cos=cos.unsqueeze(0).unsqueeze(0)
    sin=sin.unsqueeze(0).unsqueeze(0)

    return (
        x*cos+
        rotate_half(x)*sin
    )


rope_cos,rope_sin=build_rope_cache(
    cfg.max_seq_len,
    head_dim,
    cfg.rope_base
)

print("RoPE cache created")
print("cos shape:",rope_cos.shape)
print("sin shape:",rope_sin.shape)
print("head dimension:",head_dim)
```

```python
#cell 89
#RMSNorm and SwiGLU

class RMSNorm(nn.Module):

    def __init__(
        self,
        dim,
        eps=1e-6
    ):
        super().__init__()

        self.eps=eps

        self.weight=nn.Parameter(
            torch.ones(dim)
        )

    def forward(self,x):

        input_dtype=x.dtype

        x_float=x.float()

        rms=torch.rsqrt(
            x_float.pow(2).mean(
                dim=-1,
                keepdim=True
            )+
            self.eps
        )

        return (
            x_float*
            rms*
            self.weight.float()
        ).to(input_dtype)


class SwiGLU(nn.Module):

    def __init__(
        self,
        d_model,
        d_ff
    ):
        super().__init__()

        self.gate_proj=nn.Linear(
            d_model,
            d_ff,
            bias=False
        )

        self.up_proj=nn.Linear(
            d_model,
            d_ff,
            bias=False
        )

        self.down_proj=nn.Linear(
            d_ff,
            d_model,
            bias=False
        )

    def forward(self,x):

        gate=F.silu(
            self.gate_proj(x)
        )

        up=self.up_proj(x)

        return self.down_proj(
            gate*up
        )


#small sanity tests

```

```python
#scratch GQA + RoPE + local/global attention + KV cache
#optimized local attention uses chunked batched matmul
#reference local attention is retained for correctness testing


class CausalSelfAttention(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        max_seq_len,
        rope_base=10000.0,
        dropout=0.0,
        is_global=True,
        local_window=128,
        local_chunk_size=256
    ):

        super().__init__()


        assert d_model%n_heads==0

        assert n_heads%n_kv_heads==0

        assert max_seq_len>0

        assert local_window>0

        assert local_chunk_size>0


        self.d_model=d_model

        self.n_heads=n_heads

        self.n_kv_heads=n_kv_heads


        self.head_dim=(
            d_model//
            n_heads
        )


        self.n_rep=(
            n_heads//
            n_kv_heads
        )


        self.max_seq_len=max_seq_len

        self.dropout=dropout

        self.is_global=is_global

        self.local_window=local_window

        self.local_chunk_size=(
            local_chunk_size
        )


        #--------------------------------------
        #Q/K/V projections
        #--------------------------------------

        self.q_proj=nn.Linear(
            d_model,
            n_heads*self.head_dim,
            bias=False
        )


        self.k_proj=nn.Linear(
            d_model,
            n_kv_heads*self.head_dim,
            bias=False
        )


        self.v_proj=nn.Linear(
            d_model,
            n_kv_heads*self.head_dim,
            bias=False
        )


        self.out_proj=nn.Linear(
            d_model,
            d_model,
            bias=False
        )


        #--------------------------------------
        #RoPE
        #--------------------------------------

        cos,sin=build_rope_cache(
            max_seq_len,
            self.head_dim,
            rope_base
        )


        self.register_buffer(
            "rope_cos",
            cos,
            persistent=False
        )


        self.register_buffer(
            "rope_sin",
            sin,
            persistent=False
        )


    def _split_heads(
        self,
        x,
        n_heads
    ):

        B,T,_=x.shape


        return (
            x.reshape(
                B,
                T,
                n_heads,
                self.head_dim
            )
            .transpose(
                1,
                2
            )
        )


    def _apply_dropout(
        self,
        attn_weights
    ):

        if (
            self.dropout>0.0
            and
            self.training
        ):

            attn_weights=F.dropout(
                attn_weights,
                p=self.dropout
            )


        return attn_weights


    #==================================================
    #REFERENCE SCRATCH LOCAL ATTENTION
    #==================================================
    #
    #This is the original unfold/einsum implementation.
    #It is retained only for correctness comparison.
    #It is NOT used for full training.
    #

    def _local_attention_reference(
        self,
        q,
        k,
        v,
        past_len
    ):

        B,H,Dummy=q.shape[0],q.shape[1],q.shape[3]

        W=self.local_window


        #GQA expansion for reference implementation

        k=k.repeat_interleave(
            self.n_rep,
            dim=1
        )


        v=v.repeat_interleave(
            self.n_rep,
            dim=1
        )


        padded_k=F.pad(
            k,
            (
                0,
                0,
                W-1,
                0
            )
        )


        padded_v=F.pad(
            v,
            (
                0,
                0,
                W-1,
                0
            )
        )


        k_windows=(
            padded_k
            .unfold(
                2,
                W,
                1
            )
            .permute(
                0,
                1,
                2,
                4,
                3
            )
        )


        v_windows=(
            padded_v
            .unfold(
                2,
                W,
                1
            )
            .permute(
                0,
                1,
                2,
                4,
                3
            )
        )


        T=q.size(2)


        k_windows=k_windows[
            :,
            :,
            past_len:past_len+T,
            :,
            :
        ]


        v_windows=v_windows[
            :,
            :,
            past_len:past_len+T,
            :,
            :
        ]


        scores=torch.einsum(
            "bhtd,bhtwd->bhtw",
            q,
            k_windows
        )


        scores/=math.sqrt(
            self.head_dim
        )


        query_positions=(
            past_len+
            torch.arange(
                T,
                device=q.device
            )
        )


        window_offsets=torch.arange(
            W,
            device=q.device
        )


        key_positions=(
            query_positions[:,None]
            -(W-1)
            +window_offsets[None,:]
        )


        valid_mask=(
            key_positions>=0
        )


        scores=scores.masked_fill(
            ~valid_mask[
                None,
                None,
                :,
                :
            ],
            torch.finfo(
                scores.dtype
            ).min
        )


        attn_weights=F.softmax(
            scores,
            dim=-1
        )


        attn_weights=(
            self._apply_dropout(
                attn_weights
            )
        )


        out=torch.einsum(
            "bhtw,bhtwd->bhtd",
            attn_weights,
            v_windows
        )


        return out


    #==================================================
    #OPTIMIZED SCRATCH GLOBAL GQA
    #==================================================

    def _global_attention(
        self,
        q,
        k,
        v,
        past_len
    ):

        B,Hq,T,D=q.shape

        total_len=k.size(2)


        #reshape query heads into KV groups
        #
        #Q:
        #[B,Hq,T,D]
        #
        #becomes:
        #[B,Hkv,n_rep,T,D]

        q_grouped=q.reshape(
            B,
            self.n_kv_heads,
            self.n_rep,
            T,
            D
        )


        #K/V:
        #
        #[B,Hkv,K,D]
        #
        #becomes:
        #[B,Hkv,1,K,D]

        k_grouped=k.unsqueeze(2)

        v_grouped=v.unsqueeze(2)


        scores=torch.matmul(
            q_grouped,
            k_grouped.transpose(
                -2,
                -1
            )
        )


        scores/=math.sqrt(D)


        #causal mask

        causal_mask=torch.ones(
            T,
            total_len,
            dtype=torch.bool,
            device=q.device
        ).triu(
            diagonal=1+past_len
        )


        scores=scores.masked_fill(
            causal_mask[
                None,
                None,
                None,
                :,
                :
            ],
            torch.finfo(
                scores.dtype
            ).min
        )


        attn_weights=F.softmax(
            scores,
            dim=-1
        )


        attn_weights=(
            self._apply_dropout(
                attn_weights
            )
        )


        out=torch.matmul(
            attn_weights,
            v_grouped
        )


        #merge GQA groups back

        out=out.reshape(
            B,
            self.n_heads,
            T,
            D
        )


        return out


    #==================================================
    #OPTIMIZED SCRATCH LOCAL ATTENTION
    #==================================================

    def _local_attention_chunked(
        self,
        q,
        k,
        v,
        past_len
    ):

        B,Hq,T,D=q.shape

        outputs=[]


        chunk_size=(
            self.local_chunk_size
        )


        for q_start in range(
            0,
            T,
            chunk_size
        ):

            q_end=min(
                q_start+chunk_size,
                T
            )


            q_chunk=q[
                :,
                :,
                q_start:q_end,
                :
            ]


            #absolute positions

            abs_q_start=(
                past_len+
                q_start
            )

            abs_q_end=(
                past_len+
                q_end
            )


            #Only keep K/V that could possibly
            #belong to the local windows.

            k_start=max(
                0,
                abs_q_start
                -self.local_window
                +1
            )


            k_end=abs_q_end


            k_range=k[
                :,
                :,
                k_start:k_end,
                :
            ]


            v_range=v[
                :,
                :,
                k_start:k_end,
                :
            ]


            #----------------------------------
            #GQA without materializing
            #repeated K/V heads
            #----------------------------------

            q_grouped=q_chunk.reshape(
                B,
                self.n_kv_heads,
                self.n_rep,
                q_end-q_start,
                D
            )


            k_grouped=k_range.unsqueeze(
                2
            )


            v_grouped=v_range.unsqueeze(
                2
            )


            #----------------------------------
            #QK^T
            #----------------------------------

            scores=torch.matmul(
                q_grouped,
                k_grouped.transpose(
                    -2,
                    -1
                )
            )


            scores/=math.sqrt(D)


            #----------------------------------
            #local causal mask
            #----------------------------------

            query_positions=(
                torch.arange(
                    abs_q_start,
                    abs_q_end,
                    device=q.device
                )
            )


            key_positions=(
                torch.arange(
                    k_start,
                    k_end,
                    device=q.device
                )
            )


            valid_mask=(
                key_positions[None,:]
                <=
                query_positions[:,None]
            ) & (
                key_positions[None,:]
                >=
                (
                    query_positions[:,None]
                    -
                    self.local_window
                    +
                    1
                )
            )


            scores=scores.masked_fill(
                ~valid_mask[
                    None,
                    None,
                    None,
                    :,
                    :
                ],
                torch.finfo(
                    scores.dtype
                ).min
            )


            #----------------------------------
            #softmax
            #----------------------------------

            attn_weights=F.softmax(
                scores,
                dim=-1
            )


            attn_weights=(
                self._apply_dropout(
                    attn_weights
                )
            )


            #----------------------------------
            #AV
            #----------------------------------

            out_chunk=torch.matmul(
                attn_weights,
                v_grouped
            )


            #merge GQA heads

            out_chunk=out_chunk.reshape(
                B,
                self.n_heads,
                q_end-q_start,
                D
            )


            outputs.append(
                out_chunk
            )


        return torch.cat(
            outputs,
            dim=2
        )


    #==================================================
    #FORWARD
    #==================================================

    def forward(
        self,
        x,
        past_kv=None,
        use_cache=False
    ):

        B,T,_=x.shape


        past_len=(
            0
            if past_kv is None
            else past_kv[0].size(2)
        )


        if (
            past_len+T
            >
            self.max_seq_len
        ):

            raise ValueError(
                "sequence length exceeds max_seq_len"
            )


        #--------------------------------------
        #Q/K/V
        #--------------------------------------

        q=self._split_heads(
            self.q_proj(x),
            self.n_heads
        )


        k=self._split_heads(
            self.k_proj(x),
            self.n_kv_heads
        )


        v=self._split_heads(
            self.v_proj(x),
            self.n_kv_heads
        )


        #--------------------------------------
        #RoPE
        #--------------------------------------

        cos=self.rope_cos[
            past_len:
            past_len+T
        ]


        sin=self.rope_sin[
            past_len:
            past_len+T
        ]


        q=apply_rope(
            q,
            cos,
            sin
        )


        k=apply_rope(
            k,
            cos,
            sin
        )


        #--------------------------------------
        #KV cache
        #--------------------------------------

        if past_kv is not None:

            past_k,past_v=past_kv


            k=torch.cat(
                [
                    past_k,
                    k
                ],
                dim=2
            )


            v=torch.cat(
                [
                    past_v,
                    v
                ],
                dim=2
            )


        present_kv=(
            (k,v)
            if use_cache
            else None
        )


        #--------------------------------------
        #attention
        #--------------------------------------

        if self.is_global:

            out=self._global_attention(
                q,
                k,
                v,
                past_len
            )

        else:

            out=self._local_attention_chunked(
                q,
                k,
                v,
                past_len
            )


        #--------------------------------------
        #merge heads
        #--------------------------------------

        out=(
            out
            .transpose(
                1,
                2
            )
            .contiguous()
            .reshape(
                B,
                T,
                self.d_model
            )
        )


        out=self.out_proj(
            out
        )


        return out,present_kv
```

```python
#cell 92
#Pre-Norm Transformer block with local/global attention


class TransformerBlock(nn.Module):

    def __init__(
        self,
        cfg,
        is_global
    ):
        super().__init__()

        self.is_global=is_global


        self.norm1=RMSNorm(
            cfg.d_model,
            cfg.rms_norm_eps
        )


        self.attn=CausalSelfAttention(
            d_model=cfg.d_model,
            n_heads=cfg.n_heads,
            n_kv_heads=cfg.n_kv_heads,
            max_seq_len=cfg.max_seq_len,
            rope_base=cfg.rope_base,
            dropout=cfg.dropout,
            is_global=is_global,
            local_window=cfg.local_window
        )


        self.norm2=RMSNorm(
            cfg.d_model,
            cfg.rms_norm_eps
        )


        self.mlp=SwiGLU(
            cfg.d_model,
            cfg.d_ff
        )


    def forward(
        self,
        x,
        past_kv=None,
        use_cache=False
    ):

        attn_out,present_kv=self.attn(
            self.norm1(x),
            past_kv=past_kv,
            use_cache=use_cache
        )


        x=x+attn_out


        x=x+self.mlp(
            self.norm2(x)
        )


        return x,present_kv
```

```python
#cell 93
#full Legal Decoder Language Model
#local/global sliding-window architecture


class LegalDecoderLM(nn.Module):

    def __init__(
        self,
        cfg,
        gradient_checkpointing=False
    ):
        super().__init__()

        self.cfg=cfg

        self.gradient_checkpointing=(
            gradient_checkpointing
        )


        #token embedding

        self.token_embedding=nn.Embedding(
            cfg.vocab_size,
            cfg.d_model
        )


        #Transformer blocks

        self.blocks=nn.ModuleList([
            TransformerBlock(
                cfg,
                is_global=(
                    layer_idx
                    in cfg.global_layer_indices
                )
            )
            for layer_idx
            in range(cfg.n_layers)
        ])


        #final normalization

        self.final_norm=RMSNorm(
            cfg.d_model,
            cfg.rms_norm_eps
        )


        #output head

        self.lm_head=nn.Linear(
            cfg.d_model,
            cfg.vocab_size,
            bias=False
        )


        #tie input embedding and output projection

        self.lm_head.weight=(
            self.token_embedding.weight
        )


        self._init_weights()


    def _init_weights(self):

        base_std=0.02

        residual_std=(
            base_std/
            math.sqrt(
                2*self.cfg.n_layers
            )
        )


        for module in self.modules():

            if isinstance(
                module,
                nn.Linear
            ):

                if module is self.lm_head:
                    continue

                nn.init.normal_(
                    module.weight,
                    mean=0.0,
                    std=base_std
                )

                if module.bias is not None:

                    nn.init.zeros_(
                        module.bias
                    )


            elif isinstance(
                module,
                nn.Embedding
            ):

                nn.init.normal_(
                    module.weight,
                    mean=0.0,
                    std=base_std
                )


        #scale residual output projections

        for block in self.blocks:

            nn.init.normal_(
                block.attn.out_proj.weight,
                mean=0.0,
                std=residual_std
            )

            nn.init.normal_(
                block.mlp.down_proj.weight,
                mean=0.0,
                std=residual_std
            )


        #RMSNorm parameters

        for module in self.modules():

            if isinstance(
                module,
                RMSNorm
            ):

                nn.init.ones_(
                    module.weight
                )


    def forward(
        self,
        input_ids,
        past_kvs=None,
        use_cache=False
    ):

        B,T=input_ids.shape


        if T>self.cfg.max_seq_len:

            raise ValueError(
                "input sequence is longer than max_seq_len"
            )


        hidden=self.token_embedding(
            input_ids
        )


        present_kvs=[]


        for layer_idx,block in enumerate(
            self.blocks
        ):

            past_kv=None

            if past_kvs is not None:

                past_kv=past_kvs[
                    layer_idx
                ]


            if (
                self.training
                and
                self.gradient_checkpointing
                and
                not use_cache
            ):

                def block_forward(
                    hidden,
                    block=block
                ):

                    output,_=block(
                        hidden,
                        past_kv=None,
                        use_cache=False
                    )

                    return output


                hidden=torch.utils.checkpoint.checkpoint(
                    block_forward,
                    hidden,
                    use_reentrant=False
                )

                present_kv=None


            else:

                hidden,present_kv=block(
                    hidden,
                    past_kv=past_kv,
                    use_cache=use_cache
                )


            if use_cache:

                present_kvs.append(
                    present_kv
                )


        hidden=self.final_norm(
            hidden
        )


        logits=self.lm_head(
            hidden
        )


        if use_cache:

            return logits,present_kvs


        return logits
```

```python
#cell 97
#warmup + cosine learning-rate scheduler

def lr_lambda(current_step):

    if current_step < WARMUP_STEPS:

        return (
            (current_step + 1)/
            WARMUP_STEPS
        )

    progress=(
        current_step-WARMUP_STEPS
    )/(
        MAX_STEPS-WARMUP_STEPS
    )

    progress=min(
        max(progress,0.0),
        1.0
    )

    return 0.5*(
        1.0+
        math.cos(
            math.pi*progress
        )
    )


scheduler=torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda=lr_lambda
)

print("scheduler created")
print("warmup steps:",WARMUP_STEPS)
print("total steps :",MAX_STEPS)
print("initial lr  :",optimizer.param_groups[0]["lr"])
```

```python
#cell 100
#validation loss and perplexity

def evaluate_model(
    model,
    val_loader,
    device
):
    
    was_training=model.training

    model.eval()

    total_loss=torch.zeros(
        (),
        device=device
    )

    total_tokens=0

    with torch.inference_mode():

        for x,y in val_loader:

            x=x.to(
                device,
                non_blocking=True
            )

            y=y.to(
                device,
                non_blocking=True
            )

            with torch.autocast(
                device_type="cuda",
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits=model(
                    x,
                    use_cache=False
                )

                loss=F.cross_entropy(
                    logits.reshape(
                        -1,
                        cfg.vocab_size
                    ),
                    y.reshape(-1),
                    reduction="sum"
                )

            total_loss+=loss.float()

            total_tokens+=y.numel()

    mean_loss=(
        total_loss/
        total_tokens
    ).item()

    perplexity=math.exp(
        min(mean_loss,20.0)
    )

    if was_training:
        model.train()

    return mean_loss,perplexity


print("validation function created")
```

```python
#cell 102
#GPU memory benchmark for candidate micro-batch sizes

def benchmark_micro_batch(
    batch_size,
    use_checkpointing
):
    
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    
    bench_model=LegalDecoderLM(
        cfg,
        gradient_checkpointing=use_checkpointing
    ).to(device)

    bench_model.train()

    bench_params=[
        p for p in bench_model.parameters()
        if p.requires_grad
    ]

    bench_optimizer=torch.optim.AdamW(
        bench_params,
        lr=LEARNING_RATE,
        betas=(
            ADAM_BETA1,
            ADAM_BETA2
        ),
        eps=ADAM_EPS,
        weight_decay=WEIGHT_DECAY
    )

    x=torch.randint(
        0,
        cfg.vocab_size,
        (
            batch_size,
            cfg.max_seq_len
        ),
        device=device
    )

    y=torch.randint(
        0,
        cfg.vocab_size,
        (
            batch_size,
            cfg.max_seq_len
        ),
        device=device
    )

    try:

        torch.cuda.synchronize()

        start=time.perf_counter()

        bench_optimizer.zero_grad(
            set_to_none=True
        )

        with torch.autocast(
            device_type="cuda",
            dtype=AMP_DTYPE,
            enabled=USE_AMP
        ):

            logits=bench_model(
                x,
                use_cache=False
            )

            loss=F.cross_entropy(
                logits.reshape(
                    -1,
                    cfg.vocab_size
                ),
                y.reshape(-1)
            )

        scaler.benchmark if False else None

        #use a local scaler so the benchmark matches
        #the real FP16 training path
        bench_scaler=torch.amp.GradScaler(
            "cuda",
            enabled=USE_AMP
        )

        bench_scaler.scale(
            loss
        ).backward()

        bench_scaler.unscale_(
            bench_optimizer
        )

        grad_norm=torch.nn.utils.clip_grad_norm_(
            bench_model.parameters(),
            MAX_GRAD_NORM,
            error_if_nonfinite=False
        )

        old_scale=bench_scaler.get_scale()

        bench_scaler.step(
            bench_optimizer
        )

        bench_scaler.update()

        new_scale=bench_scaler.get_scale()

        torch.cuda.synchronize()

        elapsed=time.perf_counter()-start

        peak_allocated=(
            torch.cuda.max_memory_allocated()
            /1024**3
        )

        peak_reserved=(
            torch.cuda.max_memory_reserved()
            /1024**3
        )

        result={
            "batch_size":batch_size,
            "checkpointing":use_checkpointing,
            "loss":float(loss.item()),
            "grad_norm":float(grad_norm),
            "peak_allocated_gb":peak_allocated,
            "peak_reserved_gb":peak_reserved,
            "elapsed_sec":elapsed,
            "amp_scale_changed":new_scale<old_scale,
            "fits_15gb":peak_reserved<14.5
        }

    except torch.cuda.OutOfMemoryError:

        torch.cuda.empty_cache()

        result={
            "batch_size":batch_size,
            "checkpointing":use_checkpointing,
            "oom":True,
            "fits_15gb":False
        }

    finally:

        del bench_model
        del bench_optimizer
        del x
        del y

        if "logits" in locals():
            del logits

        torch.cuda.empty_cache()

    return result


print("GPU benchmark function created")
```

```python
#cell 108
#robust rng state helpers

def get_rng_state():

    rng_state={
        "python":random.getstate(),
        "numpy":np.random.get_state(),
        "torch":torch.get_rng_state().cpu().clone()
    }

    if torch.cuda.is_available():

        rng_state["cuda"]=[
            state.cpu().clone()
            for state in torch.cuda.get_rng_state_all()
        ]

    else:

        rng_state["cuda"]=None

    return rng_state


def set_rng_state(
    state
):

    if state is None:
        return

    #python rng
    random.setstate(
        state["python"]
    )

    #numpy rng
    np.random.set_state(
        state["numpy"]
    )

    #torch cpu rng
    torch_state=state["torch"]

    if not isinstance(
        torch_state,
        torch.ByteTensor
    ):

        torch_state=torch.as_tensor(
            torch_state,
            dtype=torch.uint8,
            device="cpu"
        )

    torch.set_rng_state(
        torch_state.contiguous()
    )

    #cuda rng
    if (
        torch.cuda.is_available()
        and state.get("cuda") is not None
    ):

        cuda_states=[]

        for cuda_state in state["cuda"]:

            if not isinstance(
                cuda_state,
                torch.ByteTensor
            ):

                cuda_state=torch.as_tensor(
                    cuda_state,
                    dtype=torch.uint8,
                    device="cpu"
                )

            cuda_states.append(
                cuda_state.contiguous()
            )

        torch.cuda.set_rng_state_all(
            cuda_states
        )


print(
    "robust rng helpers created"
)
```

```python
#cell 111
#deterministic cycle-aware infinite training stream

def infinite_train_batches(
    start_cycle=0,
    start_micro_batch=0
):

    cycle_index=start_cycle
    micro_batch_offset=start_micro_batch

    while True:

        cycle_dataset,cycle_loader=(
            build_train_loader(
                cycle_index
            )
        )

        #skip batches already consumed in this cycle
        loader_iterator=iter(
            cycle_loader
        )

        for _ in range(
            micro_batch_offset
        ):

            next(
                loader_iterator
            )

        for batch in loader_iterator:

            yield batch

        #cycle completed
        cycle_index+=1
        micro_batch_offset=0
```

```python
#cell 113
#final checkpoint functions for real training

from dataclasses import asdict


def save_checkpoint(
    path,
    model,
    optimizer,
    scheduler,
    scaler,
    global_step,
    tokens_seen,
    cycle_index,
    micro_batch_offset,
    best_val_loss,
    early_stop_counter
):

    checkpoint={

        "model_state_dict":
            model.state_dict(),

        "optimizer_state_dict":
            optimizer.state_dict(),

        "scheduler_state_dict":
            scheduler.state_dict(),

        "scaler_state_dict":
            (
                scaler.state_dict()
                if scaler is not None
                else None
            ),

        "global_step":
            int(global_step),

        "tokens_seen":
            int(tokens_seen),

        "cycle_index":
            int(cycle_index),

        "micro_batch_offset":
            int(micro_batch_offset),

        "best_val_loss":
            float(best_val_loss),

        "early_stop_counter":
            int(early_stop_counter),

        "rng_state":
            get_rng_state(),

        "model_config":
            asdict(cfg),

        "training_config":{

            "micro_batch_size":
                int(MICRO_BATCH_SIZE),

            "gradient_accumulation_steps":
                int(GRADIENT_ACCUMULATION_STEPS),

            "context_length":
                int(CONTEXT_LENGTH),

            "effective_tokens_per_update":
                int(EFFECTIVE_TOKENS_PER_UPDATE),

            "max_train_tokens":
                int(MAX_TRAIN_TOKENS),

            "max_steps":
                int(MAX_STEPS),

            "learning_rate":
                float(LEARNING_RATE),

            "weight_decay":
                float(WEIGHT_DECAY),

            "adam_beta1":
                float(ADAM_BETA1),

            "adam_beta2":
                float(ADAM_BETA2),

            "adam_eps":
                float(ADAM_EPS),

            "max_grad_norm":
                float(MAX_GRAD_NORM),

            "warmup_steps":
                int(WARMUP_STEPS)
        },

        "saved_at":
            time.strftime(
                "%Y-%m-%d %H:%M:%S"
            )
    }

    temp_path=path+".tmp"

    torch.save(
        checkpoint,
        temp_path
    )

    os.replace(
        temp_path,
        path
    )


def load_checkpoint(
    path,
    model,
    optimizer,
    scheduler,
    scaler
):

    checkpoint=torch.load(
        path,
        map_location=device,
        weights_only=False
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    optimizer.load_state_dict(
        checkpoint["optimizer_state_dict"]
    )

    scheduler.load_state_dict(
        checkpoint["scheduler_state_dict"]
    )

    if (
        scaler is not None
        and checkpoint.get(
            "scaler_state_dict"
        ) is not None
    ):

        scaler.load_state_dict(
            checkpoint["scaler_state_dict"]
        )

    set_rng_state(
        checkpoint["rng_state"]
    )

    return {
        "global_step":
            int(
                checkpoint["global_step"]
            ),

        "tokens_seen":
            int(
                checkpoint["tokens_seen"]
            ),

        "cycle_index":
            int(
                checkpoint["cycle_index"]
            ),

        "micro_batch_offset":
            int(
                checkpoint["micro_batch_offset"]
            ),

        "best_val_loss":
            float(
                checkpoint["best_val_loss"]
            ),

        "early_stop_counter":
            int(
                checkpoint["early_stop_counter"]
            )
    }


print("final checkpoint functions created")
```

```python
#cell 116
#final training loop

import time
import gc

#--------------------------------------------------
#training controls
#--------------------------------------------------

#run a short final-engine test first.
#after it succeeds, set this to None for the full run.
TRAINING_STEPS_OVERRIDE=100

RUN_MAX_STEPS=(
    MAX_STEPS
    if TRAINING_STEPS_OVERRIDE is None
    else min(
        TRAINING_STEPS_OVERRIDE,
        MAX_STEPS
    )
)

LOG_EVERY_STEPS=100


#--------------------------------------------------
#reproducible cycle-aware batch stream
#--------------------------------------------------

def training_batch_stream(
    start_cycle,
    start_micro_batch
):

    cycle_index=start_cycle
    micro_batch_offset=start_micro_batch

    while True:

        cycle_dataset,cycle_loader=build_train_loader(
            cycle_index
        )

        cycle_batches=len(cycle_loader)

        if micro_batch_offset>=cycle_batches:

            cycle_index+=1
            micro_batch_offset=0

            del cycle_loader
            del cycle_dataset

            gc.collect()

            continue

        loader_iterator=iter(
            cycle_loader
        )

        for batch_index,batch in enumerate(
            loader_iterator,
            start=micro_batch_offset
        ):

            next_micro_batch=(
                batch_index+1
            )

            next_cycle=cycle_index

            if next_micro_batch>=cycle_batches:

                next_cycle=(
                    cycle_index+1
                )

                next_micro_batch=0

            yield (
                batch,
                next_cycle,
                next_micro_batch
            )

        del loader_iterator
        del cycle_loader
        del cycle_dataset

        gc.collect()

        cycle_index+=1
        micro_batch_offset=0


#--------------------------------------------------
#training state
#--------------------------------------------------

model.train()

optimizer.zero_grad(
    set_to_none=True
)

batch_stream=training_batch_stream(
    cycle_index,
    micro_batch_offset
)

running_loss=0.0
running_micro_batches=0

skipped_updates=0

training_start_time=time.time()
log_window_start=time.time()
log_window_tokens=tokens_seen

last_checkpoint_time=time.time()

print("="*70)
print("STARTING TRAINING")
print("="*70)

print(
    f"starting global step      : {global_step:,}"
)

print(
    f"starting tokens seen      : {tokens_seen:,}"
)

print(
    f"starting cycle            : {cycle_index:,}"
)

print(
    f"starting micro-batch      : {micro_batch_offset:,}"
)

print(
    f"target optimizer steps    : {RUN_MAX_STEPS:,}"
)

print(
    f"tokens/update             : "
    f"{EFFECTIVE_TOKENS_PER_UPDATE:,}"
)

print(
    f"context length            : {CONTEXT_LENGTH}"
)

print(
    f"micro-batch size          : {MICRO_BATCH_SIZE}"
)

print(
    f"gradient accumulation     : "
    f"{GRADIENT_ACCUMULATION_STEPS}"
)

print(
    f"AMP dtype                 : {AMP_DTYPE}"
)

print(
    f"gradient checkpointing    : "
    f"{model.gradient_checkpointing}"
)

print("="*70)


#--------------------------------------------------
#main loop
#--------------------------------------------------

try:

    while global_step<RUN_MAX_STEPS:

        #------------------------------------------
        #one gradient accumulation window
        #------------------------------------------

        update_loss_sum=torch.zeros(
            (),
            device=device
        )

        update_micro_batches=0

        optimizer.zero_grad(
            set_to_none=True
        )

        for _ in range(
            GRADIENT_ACCUMULATION_STEPS
        ):

            (
                batch,
                cycle_index,
                micro_batch_offset
            )=next(
                batch_stream
            )

            x,y=batch

            x=x.to(
                device,
                non_blocking=True
            )

            y=y.to(
                device,
                non_blocking=True
            )

            with torch.autocast(
                device_type="cuda",
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits=model(
                    x,
                    use_cache=False
                )

                loss=F.cross_entropy(
                    logits.reshape(
                        -1,
                        cfg.vocab_size
                    ),
                    y.reshape(-1),
                    reduction="mean"
                )

                scaled_loss=(
                    loss/
                    GRADIENT_ACCUMULATION_STEPS
                )

            #--------------------------------------
            #backward
            #--------------------------------------

            if scaler is not None:

                scaler.scale(
                    scaled_loss
                ).backward()

            else:

                scaled_loss.backward()

            update_loss_sum=(
                update_loss_sum+
                loss.detach().float()
            )

            update_micro_batches+=1

            tokens_seen+=y.numel()

        #------------------------------------------
        #gradient clipping + optimizer step
        #------------------------------------------

        if scaler is not None:

            scaler.unscale_(
                optimizer
            )

        grad_norm=torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            MAX_GRAD_NORM
        )

        #------------------------------------------
        #AMP-aware optimizer step
        #------------------------------------------

        skipped=False

        if scaler is not None:

            old_scale=scaler.get_scale()

            scaler.step(
                optimizer
            )

            scaler.update()

            new_scale=scaler.get_scale()

            #a reduced scaler scale indicates
            #that the optimizer step was skipped
            skipped=(
                new_scale<old_scale
            )

        else:

            optimizer.step()

        optimizer.zero_grad(
            set_to_none=True
        )

        #------------------------------------------
        #scheduler only advances after a real
        #optimizer update
        #------------------------------------------

        if skipped:

            skipped_updates+=1

        else:

            scheduler.step()

            global_step+=1

        #------------------------------------------
        #running training statistics
        #------------------------------------------

        update_loss=(
            update_loss_sum/
            max(update_micro_batches,1)
        ).item()

        running_loss+=update_loss
        running_micro_batches+=1

        #------------------------------------------
        #periodic logging
        #------------------------------------------

        if (
            global_step>0
            and global_step%LOG_EVERY_STEPS==0
            and not skipped
        ):

            if torch.cuda.is_available():

                torch.cuda.synchronize()

            now=time.time()

            window_elapsed=(
                now-
                log_window_start
            )

            window_tokens=(
                tokens_seen-
                log_window_tokens
            )

            tokens_per_sec=(
                window_tokens/
                max(window_elapsed,1e-8)
            )

            mean_recent_loss=(
                running_loss/
                max(running_micro_batches,1)
            )

            current_lr=(
                optimizer.param_groups[0]["lr"]
            )

            elapsed_total=(
                now-
                training_start_time
            )

            print(
                f"step {global_step:>7,} | "
                f"loss {mean_recent_loss:8.4f} | "
                f"lr {current_lr:.3e} | "
                f"tokens {tokens_seen:>14,} | "
                f"grad {float(grad_norm):8.3f} | "
            )

            if torch.cuda.is_available():

                allocated=(
                    torch.cuda.memory_allocated(device)
                    /
                    (1024**3)
                )

                reserved=(
                    torch.cuda.memory_reserved(device)
                    /
                    (1024**3)
                )

                peak_allocated=(
                    torch.cuda.max_memory_allocated(device)
                    /
                    (1024**3)
                )

                peak_reserved=(
                    torch.cuda.max_memory_reserved(device)
                    /
                    (1024**3)
                )

                print(
                    f"           gpu allocated "
                    f"{allocated:.2f} GB | "
                    f"reserved {reserved:.2f} GB | "
                    f"peak allocated "
                    f"{peak_allocated:.2f} GB | "
                    f"peak reserved "
                    f"{peak_reserved:.2f} GB"
                )

            running_loss=0.0
            running_micro_batches=0

            log_window_start=now
            log_window_tokens=tokens_seen

        #------------------------------------------
        #validation
        #------------------------------------------

        should_validate=(
            global_step>0
            and global_step%VALIDATE_EVERY_STEPS==0
            and not skipped
        )

        if should_validate:

            print()
            print(
                "-"*70
            )

            print(
                f"VALIDATION AT STEP "
                f"{global_step:,}"
            )

            val_start=time.time()

            val_loss,val_ppl=evaluate_model(
                model,
                val_eval_loader,
                device
            )

            val_elapsed=time.time()-val_start

            print(
                f"validation loss          : "
                f"{val_loss:.6f}"
            )

            print(
                f"validation perplexity    : "
                f"{val_ppl:.4f}"
            )


            #--------------------------------------
            #early stopping decision
            #--------------------------------------

            improved=(
                val_loss
                <
                best_val_loss-
                EARLY_STOP_MIN_DELTA
            )

            if improved:

                best_val_loss=val_loss

                early_stop_counter=0

                save_checkpoint(
                    BEST_CHECKPOINT_PATH,
                    model,
                    optimizer,
                    scheduler,
                    scaler,
                    global_step,
                    tokens_seen,
                    cycle_index,
                    micro_batch_offset,
                    best_val_loss,
                    early_stop_counter
                )

                print(
                    "NEW BEST MODEL SAVED"
                )

                print(
                    f"best validation loss    : "
                    f"{best_val_loss:.6f}"
                )

            else:

                early_stop_counter+=1

                print(
                    f"no significant improvement | "
                    f"early-stop counter "
                    f"{early_stop_counter}/"
                    f"{EARLY_STOP_PATIENCE}"
                )

            #--------------------------------------
            #save latest state at every validation
            #so early-stop state is never lost
            #--------------------------------------

            save_checkpoint(
                LAST_CHECKPOINT_PATH,
                model,
                optimizer,
                scheduler,
                scaler,
                global_step,
                tokens_seen,
                cycle_index,
                micro_batch_offset,
                best_val_loss,
                early_stop_counter
            )

            last_checkpoint_time=time.time()

            print(
                "latest resume checkpoint saved"
            )

            print(
                "-"*70
            )
            print()

            #--------------------------------------
            #early stopping
            #--------------------------------------

            if (
                early_stop_counter
                >= EARLY_STOP_PATIENCE
            ):

                print(
                    "="*70
                )

                print(
                    "EARLY STOPPING TRIGGERED"
                )

                print(
                    f"step: {global_step:,}"
                )

                print(
                    f"best validation loss: "
                    f"{best_val_loss:.6f}"
                )

                print(
                    "="*70
                )

                break

        #------------------------------------------
        #checkpoint by step OR elapsed wall time
        #------------------------------------------

        checkpoint_by_steps=(
            global_step>0
            and
            global_step%CHECKPOINT_EVERY_STEPS==0
            and
            not skipped
        )

        checkpoint_by_time=(
            (
                time.time()-
                last_checkpoint_time
            )
            >=
            CHECKPOINT_EVERY_MINUTES*60
        )

        if (
            (
                checkpoint_by_steps
                or checkpoint_by_time
            )
            and
            not should_validate
            and
            not skipped
        ):

            save_checkpoint(
                LAST_CHECKPOINT_PATH,
                model,
                optimizer,
                scheduler,
                scaler,
                global_step,
                tokens_seen,
                cycle_index,
                micro_batch_offset,
                best_val_loss,
                early_stop_counter
            )

            last_checkpoint_time=time.time()

            reason=[]

            if checkpoint_by_steps:
                reason.append(
                    "step"
                )

            if checkpoint_by_time:
                reason.append(
                    "time"
                )

            print(
                f"checkpoint saved "
                f"({'+'.join(reason)}) | "
                f"step {global_step:,} | "
                f"tokens {tokens_seen:,} | "
                f"cycle {cycle_index:,} | "
                f"offset {micro_batch_offset:,}"
            )

    #--------------------------------------------------
    #final checkpoint
    #--------------------------------------------------

    if torch.cuda.is_available():

        torch.cuda.synchronize()

    save_checkpoint(
        LAST_CHECKPOINT_PATH,
        model,
        optimizer,
        scheduler,
        scaler,
        global_step,
        tokens_seen,
        cycle_index,
        micro_batch_offset,
        best_val_loss,
        early_stop_counter
    )

    total_elapsed=time.time()-training_start_time

    print()
    print("="*70)
    print("TRAINING FINISHED")
    print("="*70)

    print(
        f"final global step        : "
        f"{global_step:,}"
    )

    print(
        f"tokens seen              : "
        f"{tokens_seen:,}"
    )

    print(
        f"best validation loss     : "
        f"{best_val_loss:.6f}"
    )

    print(
        f"early-stop counter       : "
        f"{early_stop_counter}"
    )

    print(
        f"skipped AMP updates      : "
        f"{skipped_updates}"
    )


    print(
        f"last checkpoint exists   : "
        f"{os.path.exists(LAST_CHECKPOINT_PATH)}"
    )

    print(
        f"best checkpoint exists   : "
        f"{os.path.exists(BEST_CHECKPOINT_PATH)}"
    )

    print("="*70)


except KeyboardInterrupt:

    print()
    print(
        "KeyboardInterrupt detected."
    )

    print(
        "Saving the latest clean checkpoint..."
    )

    if torch.cuda.is_available():

        torch.cuda.synchronize()

    save_checkpoint(
        LAST_CHECKPOINT_PATH,
        model,
        optimizer,
        scheduler,
        scaler,
        global_step,
        tokens_seen,
        cycle_index,
        micro_batch_offset,
        best_val_loss,
        early_stop_counter
    )

    print(
        "checkpoint saved after interruption"
    )
```

```python
#cell 116
#full training loop

import time
import gc

#--------------------------------------------------
#full training
#--------------------------------------------------

RUN_MAX_STEPS=MAX_STEPS

LOG_EVERY_STEPS=100


#--------------------------------------------------
#cycle-aware training batch stream
#--------------------------------------------------

def training_batch_stream(
    start_cycle,
    start_micro_batch
):

    cycle_index=start_cycle
    micro_batch_offset=start_micro_batch

    while True:

        cycle_dataset,cycle_loader=build_train_loader(
            cycle_index
        )

        cycle_batches=len(cycle_loader)

        if micro_batch_offset>=cycle_batches:

            cycle_index+=1
            micro_batch_offset=0

            del cycle_loader
            del cycle_dataset

            gc.collect()

            continue

        loader_iterator=iter(
            cycle_loader
        )

        for batch_index,batch in enumerate(
            loader_iterator,
            start=micro_batch_offset
        ):

            next_micro_batch=batch_index+1
            next_cycle=cycle_index

            if next_micro_batch>=cycle_batches:

                next_cycle=cycle_index+1
                next_micro_batch=0

            yield (
                batch,
                next_cycle,
                next_micro_batch
            )

        del loader_iterator
        del cycle_loader
        del cycle_dataset

        gc.collect()

        cycle_index+=1
        micro_batch_offset=0


#--------------------------------------------------
#training state
#--------------------------------------------------

model.train()

optimizer.zero_grad(
    set_to_none=True
)

batch_stream=training_batch_stream(
    cycle_index,
    micro_batch_offset
)

running_loss=0.0
running_micro_batches=0

skipped_updates=0

training_start_time=time.time()
log_window_start=time.time()
log_window_tokens=tokens_seen

last_checkpoint_time=time.time()


print("="*70)
print("STARTING FULL TRAINING")
print("="*70)

print(
    f"starting global step      : {global_step:,}"
)

print(
    f"starting tokens seen      : {tokens_seen:,}"
)

print(
    f"starting cycle            : {cycle_index:,}"
)

print(
    f"starting micro-batch      : {micro_batch_offset:,}"
)

print(
    f"target optimizer steps    : {RUN_MAX_STEPS:,}"
)

print(
    f"tokens/update             : "
    f"{EFFECTIVE_TOKENS_PER_UPDATE:,}"
)

print(
    f"context length            : {CONTEXT_LENGTH}"
)

print(
    f"micro-batch size          : {MICRO_BATCH_SIZE}"
)

print(
    f"gradient accumulation     : "
    f"{GRADIENT_ACCUMULATION_STEPS}"
)

print(
    f"AMP dtype                 : {AMP_DTYPE}"
)

print(
    f"gradient checkpointing    : "
    f"{model.gradient_checkpointing}"
)

print("="*70)


#--------------------------------------------------
#main training loop
#--------------------------------------------------

try:

    while global_step<RUN_MAX_STEPS:

        update_loss_sum=torch.zeros(
            (),
            device=device
        )

        update_micro_batches=0

        optimizer.zero_grad(
            set_to_none=True
        )

        #------------------------------------------
        #gradient accumulation
        #------------------------------------------

        for _ in range(
            GRADIENT_ACCUMULATION_STEPS
        ):

            (
                batch,
                cycle_index,
                micro_batch_offset
            )=next(
                batch_stream
            )

            x,y=batch

            x=x.to(
                device,
                non_blocking=True
            )

            y=y.to(
                device,
                non_blocking=True
            )

            with torch.autocast(
                device_type="cuda",
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits=model(
                    x,
                    use_cache=False
                )

                loss=F.cross_entropy(
                    logits.reshape(
                        -1,
                        cfg.vocab_size
                    ),
                    y.reshape(-1),
                    reduction="mean"
                )

                scaled_loss=(
                    loss/
                    GRADIENT_ACCUMULATION_STEPS
                )

            #--------------------------------------
            #backward
            #--------------------------------------

            if scaler is not None:

                scaler.scale(
                    scaled_loss
                ).backward()

            else:

                scaled_loss.backward()

            update_loss_sum=(
                update_loss_sum+
                loss.detach().float()
            )

            update_micro_batches+=1

            tokens_seen+=y.numel()

        #------------------------------------------
        #gradient clipping
        #------------------------------------------

        if scaler is not None:

            scaler.unscale_(
                optimizer
            )

        grad_norm=torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            MAX_GRAD_NORM
        )

        #------------------------------------------
        #optimizer step
        #------------------------------------------

        skipped=False

        if scaler is not None:

            old_scale=scaler.get_scale()

            scaler.step(
                optimizer
            )

            scaler.update()

            new_scale=scaler.get_scale()

            skipped=(
                new_scale<old_scale
            )

        else:

            optimizer.step()

        optimizer.zero_grad(
            set_to_none=True
        )

        #------------------------------------------
        #scheduler
        #------------------------------------------

        if skipped:

            skipped_updates+=1

        else:

            scheduler.step()

            global_step+=1

        #------------------------------------------
        #loss statistics
        #------------------------------------------

        update_loss=(
            update_loss_sum/
            max(update_micro_batches,1)
        ).item()

        running_loss+=update_loss
        running_micro_batches+=1

        #------------------------------------------
        #logging
        #------------------------------------------

        if (
            global_step>0
            and
            global_step%LOG_EVERY_STEPS==0
            and
            not skipped
        ):

            torch.cuda.synchronize()

            now=time.time()

            window_elapsed=(
                now-
                log_window_start
            )

            window_tokens=(
                tokens_seen-
                log_window_tokens
            )

            tokens_per_sec=(
                window_tokens/
                max(window_elapsed,1e-8)
            )

            mean_recent_loss=(
                running_loss/
                max(running_micro_batches,1)
            )

            current_lr=(
                optimizer.param_groups[0]["lr"]
            )

            elapsed_total=(
                now-
                training_start_time
            )

            print(
                f"step {global_step:>7,} | "
                f"loss {mean_recent_loss:8.4f} | "
                f"lr {current_lr:.3e} | "
                f"tokens {tokens_seen:>14,} | "
                f"grad {float(grad_norm):8.3f} | "
            )

            running_loss=0.0
            running_micro_batches=0

            log_window_start=now
            log_window_tokens=tokens_seen

        #------------------------------------------
        #validation
        #------------------------------------------

        should_validate=(
            global_step>0
            and
            global_step%VALIDATE_EVERY_STEPS==0
            and
            not skipped
        )

        if should_validate:

            print()
            print("-"*70)

            print(
                f"VALIDATION AT STEP "
                f"{global_step:,}"
            )

            val_start=time.time()

            val_loss,val_ppl=evaluate_model(
                model,
                val_eval_loader,
                device
            )

            val_elapsed=time.time()-val_start

            print(
                f"validation loss       : "
                f"{val_loss:.6f}"
            )

            print(
                f"validation perplexity : "
                f"{val_ppl:.4f}"
            )


            #--------------------------------------
            #early stopping
            #--------------------------------------

            improved=(
                val_loss
                <
                best_val_loss-
                EARLY_STOP_MIN_DELTA
            )

            if improved:

                best_val_loss=val_loss
                early_stop_counter=0

                save_checkpoint(
                    BEST_CHECKPOINT_PATH,
                    model,
                    optimizer,
                    scheduler,
                    scaler,
                    global_step,
                    tokens_seen,
                    cycle_index,
                    micro_batch_offset,
                    best_val_loss,
                    early_stop_counter
                )

                print(
                    "NEW BEST MODEL SAVED"
                )

            else:

                early_stop_counter+=1

                print(
                    f"no significant improvement | "
                    f"early-stop counter "
                    f"{early_stop_counter}/"
                    f"{EARLY_STOP_PATIENCE}"
                )

            #--------------------------------------
            #latest resume checkpoint
            #--------------------------------------

            save_checkpoint(
                LAST_CHECKPOINT_PATH,
                model,
                optimizer,
                scheduler,
                scaler,
                global_step,
                tokens_seen,
                cycle_index,
                micro_batch_offset,
                best_val_loss,
                early_stop_counter
            )

            last_checkpoint_time=time.time()

            print(
                "latest resume checkpoint saved"
            )

            print("-"*70)
            print()

            #--------------------------------------
            #early stopping
            #--------------------------------------

            if (
                early_stop_counter
                >= EARLY_STOP_PATIENCE
            ):

                print("="*70)
                print("EARLY STOPPING TRIGGERED")
                print("="*70)

                break

        #------------------------------------------
        #checkpoint by time or step
        #------------------------------------------

        checkpoint_by_steps=(
            global_step>0
            and
            global_step%CHECKPOINT_EVERY_STEPS==0
            and
            not skipped
        )

        checkpoint_by_time=(
            (
                time.time()-
                last_checkpoint_time
            )
            >=
            CHECKPOINT_EVERY_MINUTES*60
        )

        if (
            (
                checkpoint_by_steps
                or
                checkpoint_by_time
            )
            and
            not should_validate
            and
            not skipped
        ):

            save_checkpoint(
                LAST_CHECKPOINT_PATH,
                model,
                optimizer,
                scheduler,
                scaler,
                global_step,
                tokens_seen,
                cycle_index,
                micro_batch_offset,
                best_val_loss,
                early_stop_counter
            )

            last_checkpoint_time=time.time()

            reason=[]

            if checkpoint_by_steps:

                reason.append(
                    "step"
                )

            if checkpoint_by_time:

                reason.append(
                    "time"
                )

            print(
                f"checkpoint saved "
                f"({' + '.join(reason)}) | "
                f"step {global_step:,} | "
                f"tokens {tokens_seen:,}"
            )

    #--------------------------------------------------
    #final checkpoint
    #--------------------------------------------------

    torch.cuda.synchronize()

    save_checkpoint(
        LAST_CHECKPOINT_PATH,
        model,
        optimizer,
        scheduler,
        scaler,
        global_step,
        tokens_seen,
        cycle_index,
        micro_batch_offset,
        best_val_loss,
        early_stop_counter
    )

    total_elapsed=(
        time.time()-
        training_start_time
    )

    print()
    print("="*70)
    print("FULL TRAINING FINISHED")
    print("="*70)

    print(
        f"final global step        : "
        f"{global_step:,}"
    )

    print(
        f"tokens seen              : "
        f"{tokens_seen:,}"
    )

    print(
        f"best validation loss     : "
        f"{best_val_loss:.6f}"
    )

    print(
        f"early-stop counter       : "
        f"{early_stop_counter}"
    )

    print(
        f"skipped AMP updates      : "
        f"{skipped_updates}"
    )


    print(
        f"last checkpoint exists   : "
        f"{os.path.exists(LAST_CHECKPOINT_PATH)}"
    )

    print(
        f"best checkpoint exists   : "
        f"{os.path.exists(BEST_CHECKPOINT_PATH)}"
    )

    print("="*70)


except KeyboardInterrupt:

    print()
    print(
        "KeyboardInterrupt detected."
    )

    print(
        "saving latest checkpoint..."
    )

    torch.cuda.synchronize()

    save_checkpoint(
        LAST_CHECKPOINT_PATH,
        model,
        optimizer,
        scheduler,
        scaler,
        global_step,
        tokens_seen,
        cycle_index,
        micro_batch_offset,
        best_val_loss,
        early_stop_counter
    )

    print(
        "checkpoint saved after interruption"
    )
```

```python
import random
import numpy as np
import torch


SPECIAL_TOKEN_IDS = {
    0,  # <pad>
    1,  # <unk>
    2,  # <s>
    3   # </s>
}


def sample_top_p(
    logits,
    temperature=1.0,
    top_p=0.95
):

    if temperature<=0:
        raise ValueError(
            "temperature must be > 0"
        )

    logits=logits/temperature

    probabilities=torch.softmax(
        logits,
        dim=-1
    )

    sorted_probs,sorted_indices=torch.sort(
        probabilities,
        descending=True,
        dim=-1
    )

    cumulative_probs=torch.cumsum(
        sorted_probs,
        dim=-1
    )

    remove_mask=(
        cumulative_probs>top_p
    )

    remove_mask[...,1:]=(
        remove_mask[...,:-1].clone()
    )

    remove_mask[...,0]=False

    sorted_probs=sorted_probs.masked_fill(
        remove_mask,
        0.0
    )

    sorted_probs=(
        sorted_probs/
        sorted_probs.sum(
            dim=-1,
            keepdim=True
        )
    )

    next_index=torch.multinomial(
        sorted_probs,
        num_samples=1
    )

    next_token=torch.gather(
        sorted_indices,
        -1,
        next_index
    )

    return next_token


def generate_ids_with_cache(
    prompt_ids,
    model,
    device,
    max_new_tokens,
    temperature,
    top_p
):

    input_ids=torch.tensor(
        [prompt_ids],
        dtype=torch.long,
        device=device
    )

    generated_ids=[]

    with torch.inference_mode():

        with torch.autocast(
            device_type="cuda",
            dtype=AMP_DTYPE,
            enabled=USE_AMP
        ):

            logits,past_kvs=model(
                input_ids,
                use_cache=True
            )

        for _ in range(max_new_tokens):

            next_logits=(
                logits[:,-1,:]
                .clone()
            )

            #never generate special tokens
            for special_id in SPECIAL_TOKEN_IDS:

                next_logits[:,special_id]=(
                    float("-inf")
                )

            next_token=sample_top_p(
                next_logits,
                temperature=temperature,
                top_p=top_p
            )

            token_id=int(
                next_token.item()
            )

            generated_ids.append(
                token_id
            )

            with torch.autocast(
                device_type="cuda",
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits,past_kvs=model(
                    next_token.view(1,1),
                    past_kvs=past_kvs,
                    use_cache=True
                )

    return generated_ids
```

```python
from collections import Counter
import math


def rouge_l_f1(
    reference_tokens,
    candidate_tokens
):

    if len(reference_tokens)==0:
        return 0.0

    if len(candidate_tokens)==0:
        return 0.0

    n=len(reference_tokens)
    m=len(candidate_tokens)

    dp=[
        [0]*(m+1)
        for _ in range(n+1)
    ]

    for i in range(1,n+1):

        for j in range(1,m+1):

            if (
                reference_tokens[i-1]
                ==
                candidate_tokens[j-1]
            ):

                dp[i][j]=(
                    dp[i-1][j-1]+1
                )

            else:

                dp[i][j]=max(
                    dp[i-1][j],
                    dp[i][j-1]
                )

    lcs_length=dp[n][m]

    precision=(
        lcs_length/
        max(len(candidate_tokens),1)
    )

    recall=(
        lcs_length/
        max(len(reference_tokens),1)
    )

    if precision+recall==0:
        return 0.0

    return (
        2*precision*recall/
        (precision+recall)
    )


def get_ngrams(tokens,n):

    return [
        tuple(tokens[i:i+n])
        for i in range(
            len(tokens)-n+1
        )
    ]


def bleu_4(
    reference_tokens,
    candidate_tokens
):

    if len(candidate_tokens)==0:
        return 0.0

    precisions=[]

    for n in range(1,5):

        reference_ngrams=Counter(
            get_ngrams(
                reference_tokens,
                n
            )
        )

        candidate_ngrams=Counter(
            get_ngrams(
                candidate_tokens,
                n
            )
        )

        if not candidate_ngrams:
            return 0.0

        clipped_count=0
        total_count=0

        for ngram,count in candidate_ngrams.items():

            clipped_count+=min(
                count,
                reference_ngrams.get(
                    ngram,
                    0
                )
            )

            total_count+=count

        precision=(
            clipped_count/
            max(total_count,1)
        )

        #smoothing so log(0) does not occur
        precision=max(
            precision,
            1e-10
        )

        precisions.append(
            precision
        )

    geometric_mean=math.exp(
        sum(
            math.log(p)
            for p in precisions
        )/4
    )

    reference_length=len(
        reference_tokens
    )

    candidate_length=len(
        candidate_tokens
    )

    if candidate_length>reference_length:

        brevity_penalty=1.0

    else:

        brevity_penalty=math.exp(
            1-
            reference_length/
            max(candidate_length,1)
        )

    return (
        brevity_penalty*
        geometric_mean
    )
```

```python
def repetition_rate(
    tokens,
    n=3
):

    if len(tokens)<n:
        return 0.0

    ngrams=[
        tuple(tokens[i:i+n])
        for i in range(
            len(tokens)-n+1
        )
    ]

    if len(ngrams)==0:
        return 0.0

    unique_ngrams=len(set(ngrams))

    return 1.0-(
        unique_ngrams/
        len(ngrams)
    )
```

```python
def apply_repetition_penalty(
    logits,
    generated_ids,
    penalty=1.1
):

    if penalty<=1.0 or len(generated_ids)==0:
        return logits

    for token_id in set(generated_ids):

        value=logits[:,token_id]

        logits[:,token_id]=torch.where(
            value<0,
            value*penalty,
            value/penalty
        )

    return logits
```

```python
def generate_ids_with_cache(
    prompt_ids,
    model,
    device,
    max_new_tokens,
    temperature,
    top_p,
    repetition_penalty=1.0
):

    input_ids=torch.tensor(
        [prompt_ids],
        dtype=torch.long,
        device=device
    )

    generated_ids=[]

    with torch.inference_mode():

        with torch.autocast(
            device_type="cuda",
            dtype=AMP_DTYPE,
            enabled=USE_AMP
        ):

            logits,past_kvs=model(
                input_ids,
                use_cache=True
            )

        for _ in range(max_new_tokens):

            next_logits=(
                logits[:,-1,:].clone()
            )

            next_logits=apply_repetition_penalty(
                next_logits,
                generated_ids,
                penalty=repetition_penalty
            )

            for special_id in SPECIAL_TOKEN_IDS:
                next_logits[:,special_id]=float("-inf")

            next_token=sample_top_p(
                next_logits,
                temperature=temperature,
                top_p=top_p
            )

            token_id=int(
                next_token.item()
            )

            generated_ids.append(token_id)

            with torch.autocast(
                device_type="cuda",
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits,past_kvs=model(
                    next_token.view(1,1),
                    past_kvs=past_kvs,
                    use_cache=True
                )

    return generated_ids
```

```python
#load final 50k A-1 BPE tokenizer

def load_a1_tokenizer(path):

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        payload=json.load(f)


    itos=payload["itos"]

    merges=[
        tuple(pair)
        for pair in payload["merges"]
    ]


    #------------------------------------------
    #base vocabulary
    #4 special tokens + 256 byte tokens
    #------------------------------------------

    id_to_bytes=[b""]*N_SPECIAL

    id_to_bytes.extend(
        bytes([b])
        for b in range(N_BYTES)
    )


    #------------------------------------------
    #reconstruct every learned BPE token
    #------------------------------------------

    for left_id,right_id in merges:

        merged_bytes=(
            id_to_bytes[left_id]
            +
            id_to_bytes[right_id]
        )

        id_to_bytes.append(
            merged_bytes
        )


    if len(itos)!=len(id_to_bytes):

        raise ValueError(
            f"itos length={len(itos)} but "
            f"id_to_bytes length={len(id_to_bytes)}"
        )


    tokenizer=A1BPETokenizer(
        itos=itos,
        id_to_bytes=id_to_bytes,
        merges=merges
    )


    if tokenizer.vocab_size!=50_000:

        raise ValueError(
            f"expected vocab size 50000, "
            f"got {tokenizer.vocab_size}"
        )


    return tokenizer


generation_tokenizer=load_a1_tokenizer(
    TOKENIZER_PATH
)


print(
    "generation tokenizer loaded"
)

print(
    "vocab size:",
    generation_tokenizer.vocab_size
)
```

```python
#top-p sampling

def sample_top_p(
    logits,
    temperature=0.7,
    top_p=0.95
):

    if temperature<=0:
        raise ValueError(
            "temperature must be > 0"
        )

    logits=logits/temperature

    probabilities=torch.softmax(
        logits,
        dim=-1
    )

    sorted_probs,sorted_indices=(
        torch.sort(
            probabilities,
            descending=True,
            dim=-1
        )
    )

    cumulative_probs=torch.cumsum(
        sorted_probs,
        dim=-1
    )

    remove_mask=(
        cumulative_probs>top_p
    )

    #keep the first token that crosses
    #the top-p threshold
    remove_mask[...,1:]=(
        remove_mask[...,:-1].clone()
    )

    remove_mask[...,0]=False

    sorted_probs=sorted_probs.masked_fill(
        remove_mask,
        0.0
    )

    sorted_probs=(
        sorted_probs/
        sorted_probs.sum(
            dim=-1,
            keepdim=True
        )
    )

    next_index=torch.multinomial(
        sorted_probs,
        num_samples=1
    )

    next_token=torch.gather(
        sorted_indices,
        -1,
        next_index
    )

    return next_token
```

```python
#repetition penalty

def apply_repetition_penalty(
    logits,
    generated_ids,
    penalty=1.10
):

    if penalty<=1.0:
        return logits

    if len(generated_ids)==0:
        return logits

    for token_id in set(
        generated_ids
    ):

        value=logits[:,token_id]

        logits[:,token_id]=torch.where(
            value<0,
            value*penalty,
            value/penalty
        )

    return logits
```

```python
#generate one continuation using KV caching

SPECIAL_TOKEN_IDS={
    0,   #<pad>
    1,   #<unk>
    2,   #<s>
    3    #</s>
}


def generate_continuation(
    prompt,
    model,
    tokenizer,
    device,
    target_words=128,
    prompt_max_tokens=224,
    max_generation_tokens=288,
    temperature=0.7,
    top_p=0.95,
    repetition_penalty=1.10
):

    #------------------------------------------
    #encode prompt
    #------------------------------------------

    prompt_ids=tokenizer.encode(
        prompt
    )


    #keep the most recent prompt tokens
    if len(prompt_ids)>prompt_max_tokens:

        prompt_ids=(
            prompt_ids[
                -prompt_max_tokens:
            ]
        )


    input_ids=torch.tensor(
        [prompt_ids],
        dtype=torch.long,
        device=device
    )


    generated_ids=[]


    #------------------------------------------
    #initial prompt forward pass
    #------------------------------------------

    with torch.inference_mode():

        with torch.autocast(
            device_type=device.type,
            dtype=AMP_DTYPE,
            enabled=USE_AMP
        ):

            logits,past_kvs=model(
                input_ids,
                use_cache=True
            )


        #--------------------------------------
        #generate one token at a time
        #--------------------------------------

        for _ in range(
            max_generation_tokens
        ):

            next_logits=(
                logits[:,-1,:].clone()
            )


            #apply repetition penalty
            next_logits=apply_repetition_penalty(
                next_logits,
                generated_ids,
                penalty=repetition_penalty
            )


            #never generate special tokens
            for special_id in SPECIAL_TOKEN_IDS:

                next_logits[
                    :,
                    special_id
                ]=-float("inf")


            #sample next token
            next_token=sample_top_p(
                next_logits,
                temperature=temperature,
                top_p=top_p
            )


            token_id=int(
                next_token.item()
            )

            generated_ids.append(
                token_id
            )


            #decode generated continuation
            generated_text=(
                tokenizer.decode(
                    generated_ids
                )
            )

            generated_words=(
                generated_text.split()
            )


            #stop exactly at 128 words
            if len(generated_words)>=target_words:

                return " ".join(
                    generated_words[
                        :target_words
                    ]
                )


            #----------------------------------
            #feed only the new token
            #----------------------------------

            with torch.autocast(
                device_type=device.type,
                dtype=AMP_DTYPE,
                enabled=USE_AMP
            ):

                logits,past_kvs=model(
                    next_token.view(1,1),
                    past_kvs=past_kvs,
                    use_cache=True
                )


    #------------------------------------------
    #max token budget reached
    #------------------------------------------

    generated_text=(
        tokenizer.decode(
            generated_ids
        )
    )

    generated_words=(
        generated_text.split()
    )

    return " ".join(
        generated_words[
            :target_words
        ]
    )
```
